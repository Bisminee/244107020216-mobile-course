```python
# Library
import cv2 as cv
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
import math
import pandas as pd
import matplotlib.pyplot as plt
from skimage import data
from collections import deque
from scipy import ndimage as ndi
from skimage import data , filters, color, segmentation, feature, morphology
from skimage.filters import gaussian
from skimage.segmentation import flood
from PIL.Image import new
```

```python
# img_binary = cv.imread('/content/drive/MyDrive/Images/gradient.jpg')
# global treshold binary
def global_treshold_binary(img, tresh):
    m, n = img.shape
    for i in range(m):
      for j in range(n):
        if img[i][j] > tresh:
          img[i][j] = 255
        else:
          img[i][j] = 0
    return img
```

```python
# binary inverse
def global_treshold_binary_inv(img, tresh):
    m, n = img.shape
    for i in range(m):
      for j in range(n):
        if img[i][j] > tresh:
          img[i][j] = 0
        else:
          img[i][j] = 255
    return img
```

```python
# tozero
def global_treshold_binary_tozero(img, tresh):
    m, n = img.shape
    for i in range(m):
      for j in range(n):
        if img[i][j] > tresh:
          img[i][j] = img[i][j]
        else:
          img[i][j] = 0
    return img
```

```python
# tozero inverse
def global_treshold_binary_tozero_inv(img, tresh):
    m, n = img.shape
    for i in range(m):
      for j in range(n):
        if img[i][j] < tresh:
          img[i][j] = img[i][j]
        else:
          img[i][j] = 0
    return img
```

```python
# trunc
def global_trunc(img, tresh):
    m, n = img.shape
    for i in range(m):
      for j in range(n):
        if img[i][j] <= tresh:
          img[i][j] = img[i][j]
        else:
          img[i][j] = tresh
    return img
```

```python
image = cv.imread('/content/drive/MyDrive/Images/gradient.jpg')
newIm = cv.cvtColor(image, cv.COLOR_BGR2GRAY)

global_treshold_binary(newIm, 200)
plt.imshow(newIm, cmap='gray')
plt.title('Global Treshold Binary')
plt.axis('off')
plt.show()
```

```python
image_inv = cv.imread('/content/drive/MyDrive/Images/gradient.jpg')
newIm_inv = cv.cvtColor(image_inv, cv.COLOR_BGR2GRAY)

global_treshold_binary_inv(newIm_inv, 200)
plt.imshow(newIm_inv, cmap='gray')
plt.title('Global Treshold Binary Inverse')
plt.axis('off')
plt.show()
```

```python
image_tozero = cv.imread('/content/drive/MyDrive/Images/gradient.jpg')
newIm_tozero = cv.cvtColor(image_tozero, cv.COLOR_BGR2GRAY)

global_treshold_binary_tozero(newIm_tozero, 200)
plt.imshow(newIm_tozero, cmap='gray')
plt.title('Global Treshold Binary tozero')
plt.axis('off')
plt.show()
```
